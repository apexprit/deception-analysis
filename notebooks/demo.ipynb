## 5. Temporal Analysis
Analyze deception probability trajectories over time.

```python
# Simulated temporal analysis for demonstration
temporal_analyzer = TemporalAnalyzer(config.temporal)

# Create sample probability trajectory
np.random.seed(42)
n_frames = 100
base_prob = np.clip(np.random.normal(0.4, 0.15, n_frames), 0, 1)
# Add spike around frame 60-70 (suspicious moment)
base_prob[55:75] = np.clip(base_prob[55:75] + 0.35, 0, 1)
timestamps = np.arange(n_frames) / config.feature.fps

# Analyze trajectory
trajectory = temporal_analyzer.analyze_trajectory(base_prob, timestamps)
spikes = temporal_analyzer.detect_spikes(base_prob, timestamps)
intervals = temporal_analyzer.find_suspicious_intervals(base_prob, timestamps)

print(f"Mean deception probability: {trajectory.get('mean_probability', np.mean(base_prob)):.3f}")
print(f"Number of spikes detected: {len(spikes)}")
print(f"Suspicious intervals: {len(intervals)}")
```

```python
# Visualize temporal analysis
temporal_viz = TemporalVisualizer()

# Create timeline plot
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(timestamps, base_prob, 'b-', alpha=0.7, label='Deception Probability')
ax.axhline(y=0.5, color='r', linestyle='--', alpha=0.5, label='Threshold')
# Highlight suspicious region
ax.axvspan(timestamps[55], timestamps[75], alpha=0.2, color='red', label='Suspicious Region')
ax.set_xlabel('Time (seconds)')
ax.set_ylabel('Deception Probability')
ax.set_title('Deception Probability Timeline')
ax.legend()
ax.set_ylim(0, 1)
plt.tight_layout()
plt.show()
```
